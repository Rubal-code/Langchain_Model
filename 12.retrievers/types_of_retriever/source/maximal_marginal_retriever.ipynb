```python
from langchain_community.vectorstores import Chroma
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
docs=[
    Document(page_content="langchain is a framework for building applications with LLMs"),
    Document(page_content="langchain provides a retriever interface to retrieve relevant documents from a vector store"),
    Document(page_content="chroma is a vector database that can be used with langchain"),
    Document(page_content="retrievers are used to retrieve relevant documents from a vector store"),
    Document(page_content="openai provides powerful embeddings model")
]
```

```python
from langchain_community.vectorstores import FAISS
embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
FAISS_vector_database=FAISS.from_documents(
    documents=docs,
    embedding=embedding_model
)
```

```python
retriever=FAISS_vector_database.as_retriever(
    search_type="mmr",
    search_kwargs={"k": 2, "fetch_k": 4, "lambda_mult": 0.5} # k = top results, lambda_mult = relevance-diversity balance
)
```

```python
query = "What is langchain?"
results = retriever.invoke(query)
```

```python
for i, doc in enumerate(results):
    print(f"\n--- Result {i+1} ---")
    print(doc.page_content)
```
