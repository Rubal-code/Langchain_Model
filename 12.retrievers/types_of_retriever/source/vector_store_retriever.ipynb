```python
from langchain_community.vectorstores import Chroma
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
embedding_model=HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
documents=[
    Document(page_content="langchain is a framework for building applications with LLMs"),
    Document(page_content="langchain provides a retriever interface to retrieve relevant documents from a vector store"),
    Document(page_content="retrievers are used to retrieve relevant documents from a vector store"),
    Document(page_content="openai provides powerful embeddings model"),
]
```

```python
vector_store=Chroma.from_documents(
    documents=documents,
    collection_name="my_collection",
    embedding=embedding_model,
    persist_directory="./chroma_db"
)
```

```python
retriever=vector_store.as_retriever(search_kwargs={"k": 2})
```

```python
query="what is retriever?"
results=retriever.invoke(query)
```

```python
for i, doc in enumerate(results):
    print(f"\n--- Result {i+1} ---")
    print(doc.page_content)
```

```python
results = vector_store.similarity_search(query, k=2)
```

```python
for i, doc in enumerate(results):
    print(f"\n--- Result {i+1} ---")
    print(doc.page_content)
```
